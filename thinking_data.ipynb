```python
import json
```

```python
def load_data(fileload):
    with open(fileload , "r") as f:
        data = json.load(f)

    return data
```

```python
data = load_data("store_data.json")
print(data)
print(type(data))
```

```python
def clear_data(data):
    str_to_int = { "one" : 1 , "two" : 2 , "three" : 3 , "four" : 4 , "five" : 5}
    for val in data:
        #clean & structured
        raw_rating = val["rating"].strip().lower()
        if (raw_rating in str_to_int):
            raw_rating = str_to_int[raw_rating]
        print(raw_rating)


data = clear_data(data)
```

```python
def getting(data):
    total_rating = 0 
    for user in data:
        total_rating += float(user["rating"])
        print(f"avg of rating = {total_rating/len(data)}")
getting(data)
```
