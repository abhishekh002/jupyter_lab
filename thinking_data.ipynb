```python
import json
```

```python
def load_data(fileload):
    with open(fileload , "r") as f:
        data = json.load(f)

    return data
```

```python
data = load_data("store_data.json")
print(data)
print(type(data))
```

```python
#extrating clear data for  json data
def clear_data(data):
    text_to_number = {"one" : 1 , "two" : 2 , "three" : 3 , "four" : 4 , "five" : 5}
    clean_user = []
    unique_val = set()
    for user in data:
        #clean raw data
        rating_review = str(user["rating"]).strip().lower()
        if rating_review in text_to_number:
            user["rating"] = float(text_to_number[rating_review])
        else:
            try:
                user["rating"] = str(rating_review)
            except ValueError:
                user["rating"] = None  # e.g. garbage/unparseable rating
        
        #hangling data
        raw_age = user.get("age")
        if (raw_age == None):
            user["age"] = None
        
        #dubication removable
        if (user["name"].strip() in unique_val):
            continue
        unique_val.add(user["name"])
        clean_user.append(user)
    return clean_user

        
```

```python
data = clear_data(data)
```

```python
#meaningful data for json
def get_insights(data):
    #avg rating
    tot_rating = 0
    for user in data:
        tot_rating += float(user["rating"])

    print(f"avg rating = {tot_rating/len(data)}")

    #poor rating
    poor_rating = 0
    for user in data:
        if (float(user["rating"]) < 3):
            poor_rating += 1
        else :
            continue
    print(f"percentage of poor rating = {poor_rating/len(data) * 100} %")
```

```python
get_insights(data)
```

```python
def get_recommendation(data):
    recommendation = []
    for user in data:
        current_recomm  = {}
        current_recomm["name"] = user["name"]
        if (float(user["rating"]) >= 4.0):
            current_recomm["brand"] = "apple"
        else :
            current_recomm["brand"] = "sumsung"

        recommendation.append(current_recomm)
    return recommendation
            
            
    
```

```python
get_recommendation(data)
```
