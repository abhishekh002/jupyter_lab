```python
import pandas as pd
```

```python
df = pd.read_csv("Salary_Dataset_DataScienceLovers.csv")
```

```python
df2 = df.copy()
```

```python
new_rating_oder = [col for col in df2.columns if col != "Rating"] + ["Rating"]
df3 = df2[new_rating_oder].sort_values(["Rating"])
df3
```

```python
df3.groupby("Company Name")["Salary"].mean().sort_values(ascending=False).reset_index()
```

```python
df4 = df3.groupby("Job Title").agg(
    Max_Salary = ("Salary", "max"),
    jobsatifaction = ("Rating", "min")).reset_index()
```

```python
df4.to_json("new_salary.json")
```

```python
df4.to_excel("demo.xlsx")
```
