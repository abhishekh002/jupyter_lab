```python
import pandas as pd
```

```python
df = pd.read_csv("new_salary.csv")
```

```python
df
```

```python
df_long = df.melt(
    id_vars=["Job Title"], 
    value_vars=["Max_Salary", "jobsatifaction"],
    var_name="Metric",
    value_name="Value"
)
```

```python
df_long
```

```python
df_long.to_csv("melt.csv")
```

```python
df_wide = df_long.pivot(
    index ="Job Title",
    columns = "Metric",
    values = "Value").reset_index()
```

```python
df_wide
```

```python
df_wide.hist(["jobsatifaction"])
```

```python
df_wide.plot(kind = "scatter" , y = "Max_Salary" , x = "jobsatifaction")
```

```python
df1 = pd.DataFrame({
    "Name" : ["abhishekh" , "rohan" , "bob"],
    "Roll_no" : [ 11 , 128 , 10]
    })
df2 = pd.DataFrame({
    "Roll_no" : [ 11 , 128 , 10],
    "mark" : [ 25 , 77 , "f"]
        })
```
