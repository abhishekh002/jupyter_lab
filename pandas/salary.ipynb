```python
import pandas as pd
```

```python
df = pd.read_csv("Salary.csv")
df
```

```python
df2 = df.copy()
df2.columns
```

```python
df2["YearsExperience"] = df2["YearsExperience"].apply(lambda x:"20% increment" if x >= 5  else "freshers")
```

```python
df2
```

```python
maping = { "freshers" : "New in job " , "20% increment" : "experience" }
df2["YearsExperience"] = df2["YearsExperience"].map(maping)
df2
```

```python
df2 = df2.assign(year = df2["YearsExperience"])
df2
```

```python
df2["year"] = df2["year"].replace(["New in job ", "experience"],[1,2])
df2
```

```python
df2 = df2.sort_values(by= "Salary")
df2
```

```python
sorted_df = df2.sort_values(by = "Salary" , ascending = False)
```

```python
sorted_df
```

```python
a_df = sorted_df.reset_index(drop = True)
a_df
```

```python
a_df["ranking"] = a_df["Salary"].rank()
a_df.sort_values(by = "ranking")
```
